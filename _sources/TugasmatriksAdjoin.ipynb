# SOAL MATRIKD ADJOIN

# 1. Matriks 5×5

Diketahui matriks:

$$
\small
A =
\left[
\begin{array}{ccccc}
a & b & c & d & e\\
f & g & h & i & j\\
k & l & m & n & o\\
p & q & r & s & t\\
u & v & w & x & y
\end{array}
\right]
$$

Matriks di atas digunakan sebagai dasar dalam proses pencarian minor, kofaktor, determinan, adjoin, dan invers matriks.

# 2. Minor Matriks

Minor adalah determinan submatriks yang diperoleh dengan menghapus satu baris dan satu kolom tertentu dari matriks asal.

Rumus minor:

$$
M_{ij}
$$

Contoh minor pada baris pertama:

$$
\small
M_{11}=
\left|
\begin{array}{cccc}
g & h & i & j\\
l & m & n & o\\
q & r & s & t\\
v & w & x & y
\end{array}
\right|
$$

$$
\small
M_{12}=
\left|
\begin{array}{cccc}
f & h & i & j\\
k & m & n & o\\
p & r & s & t\\
u & w & x & y
\end{array}
\right|
$$

$$
\small
M_{13}=
\left|
\begin{array}{cccc}
f & g & i & j\\
k & l & n & o\\
p & q & s & t\\
u & v & x & y
\end{array}
\right|
$$

$$
\small
M_{14}=
\left|
\begin{array}{cccc}
f & g & h & j\\
k & l & m & o\\
p & q & r & t\\
u & v & w & y
\end{array}
\right|
$$

$$
\small
M_{15}=
\left|
\begin{array}{cccc}
f & g & h & i\\
k & l & m & n\\
p & q & r & s\\
u & v & w & x
\end{array}
\right|
$$

# 3. Kofaktor Matriks

Kofaktor diperoleh dari minor yang dikalikan dengan tanda positif atau negatif sesuai posisinya.

$$
C_{ij}=(-1)^{i+j}M_{ij}
$$

Pola tanda kofaktor:

$$
\small
\left[
\begin{array}{ccccc}
+ & - & + & - & +\\
- & + & - & + & -\\
+ & - & + & - & +\\
- & + & - & + & -\\
+ & - & + & - & +
\end{array}
\right]
$$

Contoh kofaktor pada baris pertama:

$$
C_{11}=+M_{11}
$$

$$
C_{12}=-M_{12}
$$

$$
C_{13}=+M_{13}
$$

$$
C_{14}=-M_{14}
$$

$$
C_{15}=+M_{15}
$$

$$
\small
C=
\left[
\begin{array}{ccccc}
C_{11} & C_{12} & C_{13} & C_{14} & C_{15}\\
C_{21} & C_{22} & C_{23} & C_{24} & C_{25}\\
C_{31} & C_{32} & C_{33} & C_{34} & C_{35}\\
C_{41} & C_{42} & C_{43} & C_{44} & C_{45}\\
C_{51} & C_{52} & C_{53} & C_{54} & C_{55}
\end{array}
\right]
$$

# 4. Determinan Matriks

Determinan matriks 5×5 dapat dihitung menggunakan ekspansi kofaktor pada baris pertama.

$$
det(A)=aC_{11}+bC_{12}+cC_{13}+dC_{14}+eC_{15}
$$

Karena tanda kofaktor pada baris pertama adalah:

$$
(+,-,+,-,+)
$$

maka diperoleh:

$$
det(A)=aM_{11}-bM_{12}+cM_{13}-dM_{14}+eM_{15}
$$

Artinya setiap elemen pada baris pertama dikalikan dengan minor yang sesuai kemudian dijumlahkan berdasarkan tanda kofaktornya.

# 5. Ekspansi Kofaktor

## Bagian $aM_{11}$

$$
\small
aM_{11}
=
a
\left|
\begin{array}{cccc}
g & h & i & j\\
l & m & n & o\\
q & r & s & t\\
v & w & x & y
\end{array}
\right|
$$

Ekspansi menjadi determinan 3×3:

$$
\small
aM_{11}
=
a
\left[
g
\left|
\begin{array}{ccc}

r & s & t\\
w & x & y
\end{array}
\right|
-
h
\left|
\begin{array}{ccc}
l & n & o\\

\end{array}
\right|
+
i
\left|
\begin{array}{ccc}
l & m & o\\

\end{array}
\right|
-

\left|
\begin{array}{ccc}
l & m & n\\
q & r & s\\
v & w & x
\end{array}
\right|
\right]
$$

Salah satu determinan 3×3:

$$
\small
\left|
\begin{array}{ccc}
m & n & o\\
r & s & t\\

\end{array}
\right|
=
m(sy-tx)-n(ry-tw)+o(rx-sw)
$$

Sehingga:

$$
\small
aM_{11}
=
a[

$$

$$
\small
-h(l(sy-tx)-n(qy-tv)+o(qx-sv))
$$

$$
\small
+i(l(ry-tw)-m(qy-tv)+o(qw-rv))
$$

$$
\small
-j(l(rx-sw)-m(qx-sv)+n(qw-rv))
]
$$

Untuk $M_{12}$, $M_{13}$, $M_{14}$, dan $M_{15}$ proses ekspansinya dilakukan dengan langkah yang sama seperti pada $M_{11}$.

# 6. Determinan Akhir

Setelah seluruh ekspansi selesai dilakukan, diperoleh:

$$
det(A)=aM_{11}-bM_{12}+cM_{13}-dM_{14}+eM_{15}
$$

Nilai tersebut merupakan determinan dari matriks $A$.

# 7. Adjoin Matriks

Adjoin adalah transpose dari matriks kofaktor.

$$
Adj(A)=C^T
$$

Jika matriks kofaktor:

$$
\small
C=
\left[
\begin{array}{ccccc}
C_{11} & C_{12} & C_{13} & C_{14} & C_{15}\\
C_{21} & C_{22} & C_{23} & C_{24} & C_{25}\\
C_{31} & C_{32} & C_{33} & C_{34} & C_{35}\\
C_{41} & C_{42} & C_{43} & C_{44} & C_{45}\\
C_{51} & C_{52} & C_{53} & C_{54} & C_{55}
\end{array}
\right]
$$

Maka:

$$
\small
Adj(A)=
\left[
\begin{array}{ccccc}
C_{11} & C_{21} & C_{31} & C_{41} & C_{51}\\
C_{12} & C_{22} & C_{32} & C_{42} & C_{52}\\
C_{13} & C_{23} & C_{33} & C_{43} & C_{53}\\
C_{14} & C_{24} & C_{34} & C_{44} & C_{54}\\
C_{15} & C_{25} & C_{35} & C_{45} & C_{55}
\end{array}
\right]
$$

Transpose berarti baris menjadi kolom dan kolom menjadi baris.

# 8. Invers Matriks

Invers matriks diperoleh dengan rumus:

$$
A^{-1}
=
\frac{1}{det(A)}
Adj(A)
$$

Syarat matriks memiliki invers:

$$
det(A)\neq0
$$

Jika determinan sama dengan nol, maka matriks tidak memiliki invers.

$$
\small
A^{-1}
=
\frac{1}{det(A)}
\left[
\begin{array}{ccccc}
C_{11} & C_{21} & C_{31} & C_{41} & C_{51}\\
C_{12} & C_{22} & C_{32} & C_{42} & C_{52}\\
C_{13} & C_{23} & C_{33} & C_{43} & C_{53}\\
C_{14} & C_{24} & C_{34} & C_{44} & C_{54}\\
C_{15} & C_{25} & C_{35} & C_{45} & C_{55}
\end{array}
\right]
$$

# 9. Contoh Adjoin Matriks

Diketahui matriks kofaktor:

$$
C=
\left[
\begin{array}{ccc}
1&2&3\\
4&5&6\\
7&8&9
\end{array}
\right]
$$

Adjoin diperoleh dengan melakukan transpose pada matriks kofaktor.

$$
Adj(A)=C^T
$$

Maka:

$$
Adj(A)=
\left[
\begin{array}{ccc}
1&4&7\\
2&5&8\\
3&6&9
\end{array}
\right]
$$

Baris pada matriks kofaktor berubah menjadi kolom pada matriks adjoin.

# 10. Contoh Invers Matriks

Diketahui:

$$
det(A)=5
$$

dan:

$$
Adj(A)=
\left[
\begin{array}{cc}
2&1\\
3&4
\end{array}
\right]
$$

Rumus invers:

$$
A^{-1}
=
\frac{1}{det(A)}
Adj(A)
$$

Substitusi nilai:

$$
A^{-1}
=
\frac{1}{5}
\left[
\begin{array}{cc}
2&1\\
3&4
\end{array}
\right]
$$

Hasil:

$$
A^{-1}
=
\left[
\begin{array}{cc}
0.4&0.2\\
0.6&0.8
\end{array}
\right]
$$

# 11. Contoh Perhitungan Determinan Matriks 3×3

Diketahui:

$$
A=
\left[
\begin{array}{ccc}
1&2&3\\
0&1&4\\
5&6&0
\end{array}
\right]
$$

Menggunakan ekspansi kofaktor baris pertama:

$$
det(A)=1C_{11}+2C_{12}+3C_{13}
$$

Mencari $C_{11}$:

$$
M_{11}
=
\left|
\begin{array}{cc}
1&4\\
6&0
\end{array}
\right|
$$

$$
M_{11}=(1\times0)-(4\times6)
$$

$$
M_{11}=-24
$$

Karena posisi $C_{11}$ bertanda positif:

$$
C_{11}=-24
$$


Mencari $C_{12}$:

$$
M_{12}
=
\left|
\begin{array}{cc}
0&4\\
5&0
\end{array}
\right|
$$

$$
M_{12}=(0\times0)-(4\times5)
$$

$$
M_{12}=-20
$$

Karena posisi $C_{12}$ bertanda negatif:

$$
C_{12}=-(-20)
$$

$$
C_{12}=20
$$


Mencari $C_{13}$:

$$
M_{13}
=
\left|
\begin{array}{cc}
0&1\\
5&6
\end{array}
\right|
$$

$$
M_{13}=(0\times6)-(1\times5)
$$

$$
M_{13}=-5
$$

Karena posisi $C_{13}$ bertanda positif:

$$
C_{13}=-5
$$


Substitusi ke rumus determinan:

$$
det(A)=1(-24)+2(20)+3(-5)
$$

$$
det(A)=-24+40-15
$$

$$
det(A)=1
$$

Jadi nilai determinan matriks tersebut adalah:

$$
det(A)=1
$$
