# Tugas EigenValue dan EigenFaktor

## QR Decompotition

Matriks yang digunakan:

$$
A =
\begin{bmatrix}
2 & 1 \\
1 & 2
\end{bmatrix}
$$

# Langkah-Langkah
## 1. Membentuk Vektor dari Matriks
Kolom pertama matriks dijadikan vektor $A_1$ dan kolom kedua dijadikan vektor $A_2$.

$$
A_1 =
\begin{bmatrix}
2 \\
1
\end{bmatrix}
\quad
A_2 =
\begin{bmatrix}
1 \\
2
\end{bmatrix}
$$


## 2. Menentukan $v_1$
Pada metode Gram-Schmidt:

$$
v_1 = A_1
$$

Sehingga:

$$
v_1 =
\begin{bmatrix}
2 \\
1
\end{bmatrix}
$$


## 3. Menghitung Panjang Vektor $I_1$
Panjang vektor dihitung menggunakan rumus:

$$
||v|| = \sqrt{x^2 + y^2}
$$

Maka:

$$
I_1 = \sqrt{2^2 + 1^2} = \sqrt5
$$


## 4. Menghitung $q_1$
Vektor $q_1$ diperoleh dari normalisasi vektor $v_1$.

$$
q_1 = \frac{v_1}{I_1}
$$

Hasil:

$$
q_1 =
\begin{bmatrix}
2/\sqrt5 \\
1/\sqrt5
\end{bmatrix}
$$


## 5. Menghitung Dot Product
Dot product digunakan untuk mencari proyeksi vektor.

$$
q_1 \cdot A_2
$$

Hasil:

$$
q_1 \cdot A_2 = 4/\sqrt5
$$


## 6. Menghitung Proyeksi
Proyeksi dihitung dengan rumus:

$$
(q_1 \cdot A_2)q_1
$$

Hasil:

$$
\begin{bmatrix}
8/5 \\
4/5
\end{bmatrix}
$$


## 7. Menghitung $v_2$
Vektor kedua dihitung menggunakan:

$$
v_2 = A_2 - (q_1 \cdot A_2)q_1
$$

Hasil:

$$
v_2 =
\begin{bmatrix}
-3/5 \\
6/5
\end{bmatrix}
$$


## 8. Menghitung $I_2$
Panjang vektor kedua:

$$
I_2 = ||v_2||
$$

Hasil:

$$
I_2 = 3/\sqrt5
$$


## 9. Menghitung $q_2$
Normalisasi vektor kedua:

$$
q_2 = \frac{v_2}{I_2}
$$

Hasil:

$$
q_2 =
\begin{bmatrix}
-1/\sqrt5 \\
2/\sqrt5
\end{bmatrix}
$$


## 10. Membentuk Matriks $Q$
Matriks $Q$ dibentuk dari vektor $q_1$ dan $q_2$.

$$
Q =
\begin{bmatrix}
2/\sqrt5 & -1/\sqrt5 \\
1/\sqrt5 & 2/\sqrt5
\end{bmatrix}
$$


## 11. Menghitung Transpose $Q^T$
Transpose matriks digunakan untuk mencari matriks $R$.

$$
Q^T =
\begin{bmatrix}
2/\sqrt5 & 1/\sqrt5 \\
-1/\sqrt5 & 2/\sqrt5
\end{bmatrix}
$$


## 12. Menghitung Matriks $R$
Matriks $R$ dihitung menggunakan:

$$
R = Q^T A
$$

Hasil:

$$
R =
\begin{bmatrix}
\sqrt5 & 4/\sqrt5 \\
0 & 3/\sqrt5
\end{bmatrix}
$$


# Iterasi QR

## 13. Iterasi QR hingga $A_{10}$
Setelah mendapatkan matriks $Q$ dan $R$, dilakukan iterasi QR menggunakan rumus:

$$
A_{k+1} = RQ
$$

Langkah iterasi:
1. Matriks $A_k$ difaktorkan menjadi $Q_k$ dan $R_k$
2. Dibentuk matriks baru:

$$
A_{k+1} = R_kQ_k
$$

3. Proses diulang hingga iterasi ke-10.


# Kesimpulan
Berdasarkan hasil perhitungan QR Decomposition dan Iterasi QR pada matriks $2 \times 2$, diperoleh matriks ortogonal $Q$ dan matriks segitiga atas $R$. Selanjutnya dilakukan proses iterasi QR hingga iterasi ke-10 menggunakan rumus $A_{k+1} = RQ$. Dari hasil iterasi tersebut, matriks secara bertahap mendekati bentuk matriks diagonal, dimana elemen di bawah diagonal semakin kecil mendekati nol. Nilai diagonal matriks hasil iterasi mendekati eigenvalue dari matriks awal, yaitu $\lambda_1 = 3$ dan $\lambda_2 = 1$. Dengan demikian, metode QR Iteration dapat digunakan untuk mencari pendekatan nilai eigen melalui proses iterasi matriks secara berulang.

## Kode Program

```python
import math
from fractions import Fraction
import sympy as sp

# Matriks 2x2
A = [
    [2, 1],
    [1, 2]
]

print("=== Matriks A ===")
for baris in A:
    print(baris)

# Mengambil kolom pertama dan kedua
A1 = [A[0][0], A[1][0]]
A2 = [A[0][1], A[1][1]]

print("\n=== Langkah 1: Membentuk Vektor ===")
print("Vektor pertama (A1) =", A1)
print("Vektor kedua (A2) =", A2)

# Menentukan v1
v1 = A1

print("\n=== Langkah 2: Menentukan v1 ===")
print("v1 = A1")
print("v1 =", v1)

# Menghitung I1
I1 = math.sqrt(v1[0]**2 + v1[1]**2)

print("\n=== Langkah 3: Menghitung I1 ===")
print("I1 = √(2² + 1²)")
print("I1 = √5")

# Menghitung q1
q1 = [
    v1[0] / I1,
    v1[1] / I1
]

print("\n=== Langkah 4: Menghitung q1 ===")
print("q1 = v1 / I1")
print("q1 = [2/√5, 1/√5]")

# Dot product q1 . A2
dot = q1[0]*A2[0] + q1[1]*A2[1]

print("\n=== Langkah 5: Menghitung Dot Product ===")
print("q1 . A2 = 4/√5")

# Menghitung proyeksi
hasil = [
    dot * q1[0],
    dot * q1[1]
]

hasil_pecahan = [
    Fraction(hasil[0]).limit_denominator(),
    Fraction(hasil[1]).limit_denominator()
]

print("\n=== Langkah 6: Menghitung Proyeksi ===")
print("(q1 . A2)q1 = [",
      f"{hasil_pecahan[0].numerator}/{hasil_pecahan[0].denominator},",
      f"{hasil_pecahan[1].numerator}/{hasil_pecahan[1].denominator}",
      "]")

# Menghitung v2
v2 = [
    A2[0] - hasil[0],
    A2[1] - hasil[1]
]

v2_pecahan = [
    Fraction(v2[0]).limit_denominator(),
    Fraction(v2[1]).limit_denominator()
]

print("\n=== Langkah 7: Menghitung v2 ===")
print("v2 = A2 - ((q1 . A2)q1)")
print("v2 = [",
      f"{v2_pecahan[0].numerator}/{v2_pecahan[0].denominator},",
      f"{v2_pecahan[1].numerator}/{v2_pecahan[1].denominator}",
      "]")

# Menghitung I2
I2 = math.sqrt(v2[0]**2 + v2[1]**2)

print("\n=== Langkah 8: Menghitung I2 ===")
print("I2 = ||v2||")
print("I2 = 3/√5")

# Menghitung q2
q2 = [
    v2[0] / I2,
    v2[1] / I2
]

print("\n=== Langkah 9: Menghitung q2 ===")
print("q2 = v2 / I2")
print("q2 = [-1/√5, 2/√5]")

# Membentuk matriks Q
print("\n=== Langkah 10: Membentuk Matriks Q ===")
print("Rumus : Q = [ q1 q2 ]")

print("Q =")
for baris in Q.tolist():
    print(str(baris).replace("sqrt(5)", "√5"))

print("\n=== Langkah 11: Transpose Q ===")
print("Rumus : Q^T")

print("Q^T =")
for baris in QT.tolist():
    print(str(baris).replace("sqrt(5)", "√5"))

print("\n=== Langkah 12: Menghitung Matriks R ===")
print("Rumus : R = Q^T A")

print("R =")
for baris in R.tolist():
    print(str(baris).replace("sqrt(5)", "√5"))

print("\n=== Hasil Akhir QR Decomposition ===")
print("A = QR")

print("\n=== Langkah 13: Iterasi QR ===")

# Mengubah matriks A ke bentuk sympy
A_iterasi = sp.Matrix(A)

# Iterasi sampai A10
for i in range(1, 11):

    print(f"\n========== Iterasi ke-{i} ==========")

    # QR Decomposition
    Q_iterasi, R_iterasi = A_iterasi.QRdecomposition()

    print("\nMatriks Q:")
    sp.pprint(Q_iterasi.evalf(4))

    print("\nMatriks R:")
    sp.pprint(R_iterasi.evalf(4))

    # Menghitung A baru
    A_iterasi = sp.simplify(R_iterasi * Q_iterasi)

    print(f"\nA{i} = RQ")
    sp.pprint(A_iterasi.evalf(4))
```
